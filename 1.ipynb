```python
import numpy as np
import cv2

def generate_salt_noise(image):
    num_salt = np.ceil(0.05 * image.size)
    coords = [np.random.randint(0, i-1, int(num_salt)) for i in image.shape]
    salted_image = image.copy()
    salted_image[coords[0], coords[1]] = 255
    return salted_image
```

```python
def generate_pepper_noise(image):
    num_pepper = np.ceil(0.05 * image.size)
    coords = [np.random.randint(0, i-1, int(num_pepper)) for i in image.shape]
    peppered_image = image.copy()
    peppered_image[coords[0], coords[1]] = 0
    return peppered_image
    
```

```python
lenna_image = cv2.imread('./data/like_lenna.png', cv2.IMREAD_GRAYSCALE)

salted_lenna = generate_salt_noise(lenna_image)
pepper_lenna = generate_pepper_noise(salted_lenna)
filtered_lenna = cv2.medianBlur(pepper_lenna, 5)

fig, axes = plt.subplots(1, 4, figsize=(20, 6))

axes[0].imshow(lenna_image, cmap= 'gray')
axes[1].imshow(salted_lenna, cmap= 'gray')
axes[2].imshow(pepper_lenna, cmap= 'gray')
axes[3].imshow(filtered_lenna, cmap= 'gray')
plt.tight_layout()
plt.show()
```

```python
image = cv2.imread('./data/like_lenna.png', cv2. IMREAD_GRAYSCALE)
mean = 0
sigma = 1
gaussian_noise = np.random.normal(mean, sigma, image.shape).astype('uint8')
noisy_image = cv2.add(image, gaussian_noise)
plt.imshow(noisy_image, cmap= 'gray')
plt.show()
```

```python
sigma_values = [1, 5, 10]
denoised_images = []

for sigma in sigma_values:
    denoised = cv2.GaussianBlur(noisy_image, (0, 0), sigma)
    denoised_images.append(denoised)

fig, axes = plt.subplots(1, 4, figsize = (20, 10))

axes[0].imshow(noisy_image, cmap= 'gray')

for ax, img, sigma in zip(axes[1:], denoised_images, sigma_values):
    ax.imshow(img, cmap= 'gray')

plt.tight_layout()
plt.show()
```

```python
img = cv2.imread('./data/like_lenna.png')
img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
plt.imshow(img)
plt.show()
```

```python
def rotate_image(image, angle, center = None) :
    rows, cols, _ = image.shape
    if center is None :
        center = (cols // 2, rows // 2)

    M = cv2.getRotationMatrix2D(center, angle, 1)
    rotated = cv2.warpAffine(image, M, (cols, rows))
    return rotated

rotated_img = rotate_image(img, 45)
plt.imshow(rotated_img)
plt.show()
```

```python
image_path = './data/perspective_test.jpg'
new_source_image = cv2.imread(image_path)

ordered_coners = np.array([[57, 640], [936, 330], [1404, 792], [550, 1431]], dtype= 'float32')
ordered_width = int(max(np.linalg.norm(ordered_coners[0] - ordered_coners[1]),
                    np.linalg.norm(ordered_coners[2] - ordered_coners[3])))
ordered_height = int(max(np.linalg.norm(ordered_coners[0] - ordered_coners[3]),
                    np.linalg.norm(ordered_coners[1] - ordered_coners[2])))

ordered_rect_coners = np.array([[0,0], [ordered_width, 0], [ordered_width, ordered_height],
                               [0, ordered_height]], dtype= 'float32')

ordered_scan_matrix = cv2.getPerspectiveTransform(ordered_coners, ordered_rect_coners)
ordered_scanned_image = cv2.warpPerspective(new_source_image, ordered_scan_matrix, (ordered_width, ordered_height))

plt.figure(figsize=(10, 5))
plt.subplot(1, 2, 1)
plt.imshow(cv2.cvtColor(new_source_image, cv2.COLOR_BGR2RGB))
plt.subplot(1, 2, 2)
plt.imshow(cv2.cvtColor(ordered_scanned_image, cv2.COLOR_BGR2RGB))
plt.show()
```
